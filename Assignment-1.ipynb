```python
import math

# 1. Integer (int)
a = 10
b = 3
print("1. INTEGER")
print("a =", a, "Type:", type(a))
print("Addition:", a + b)
print("Subtraction:", a - b)
print("Multiplication:", a * b)
print("Division:", a / b)
print("Floor Division:", a // b)
print("Power:", a ** b)
```

```python
# 2. Float
x = 12.5
print("\n2. FLOAT")
print("Value:", x, "Type:", type(x))
print("Round:", round(x))
print("Absolute:", abs(-x))
```

```python
# 3. String (str)
s = "Python Programming"
print("\n4. STRING")
print("String:", s, "Type:", type(s))
print("Uppercase:", s.upper())
print("Lowercase:", s.lower())
print("Length:", len(s))
print("Replace:", s.replace("Python", "Java"))
print("First character:", s[0])
```

```python
# 4. List
lst = [10, 20, 30, 40]
print("\n5. LIST")
print("List:", lst, "Type:", type(lst))
lst.append(50)
print("After append:", lst)
lst.remove(20)
print("After remove:", lst)
print("Length:", len(lst))
print("First element:", lst[0])
```

```python
# 5. Tuple
t = (10, 20, 30, 20)
print("\n6. TUPLE")
print("Tuple:", t, "Type:", type(t))
print("Length:", len(t))
print("Count of 20:", t.count(20))
print("Index of 30:", t.index(30))
print("First element:", t[0])
```

```python
# 6. Dictionary (dict)
student = {"name": "Raj", "age": 21, "marks": 85}
print("\n7. DICTIONARY")
print("Dictionary:", student, "Type:", type(student))
print("Student name:", student["name"])
student["grade"] = "A"
print("After adding grade:", student)
print("Keys:", list(student.keys()))
print("Values:", list(student.values()))
print("Items:", list(student.items()))
```

```python
# 7. Set
st = {10, 20, 30, 40}
st2 = {30, 40, 50, 60}
print("\n8. SET")
print("Set:", st, "Type:", type(st))
print("Union:", st | st2)
print("Intersection:", st & st2)
print("Difference:", st - st2)
st.add(50)
print("After add:", st)
st.discard(10)
print("After discard:", st)
```

```python
# 8. Frozenset
fs = frozenset([1, 2, 3, 4])
print("\n9. FROZENSET")
print("Frozenset:", fs, "Type:", type(fs))
print("Length:", len(fs))
print("Contains 2:", 2 in fs)
print("Union:", fs.union({4, 5, 6}))
print("Intersection:", fs.intersection({2, 3, 8}))
```

```python
# 9. Boolean (bool)
p = True
q = False
print("\n10. BOOLEAN")
print("p =", p, "Type:", type(p))
print("AND:", p and q)
print("OR:", p or q)
print("NOT:", not p)
print("Comparison:", 10 > 5)
```

```python
# Built-in functions
print("\n12. BUILT-IN FUNCTIONS")
numbers = [5, 10, 15, 20]
print("Length:", len(numbers))
print("Maximum:", max(numbers))
print("Minimum:", min(numbers))
print("Sum:", sum(numbers))
print("Sorted:", sorted(numbers))
print("Integer conversion:", int(5.8))
print("Float conversion:", float(10))
print("String conversion:", str(123))
print("Boolean conversion:", bool(1))
```
