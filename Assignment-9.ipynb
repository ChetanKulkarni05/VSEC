```python
import seaborn as sns
import matplotlib.pyplot as plt
import pandas as pd

df = pd.DataFrame({
    "day": ["Mon", "Tue", "Wed", "Thu", "Fri"],
    "bill": [100, 200, 150, 250, 180],
    "tip": [10, 20, 15, 25, 18]
})

sns.lineplot(data=df, x="day", y="bill")
plt.show()

sns.barplot(data=df, x="day", y="bill")
plt.show()

sns.histplot(data=df, x="bill")
plt.show()

sns.boxplot(data=df, y="bill")
plt.show()

sns.scatterplot(data=df, x="bill", y="tip")
plt.show()
```
