```python
import numpy as np

# Define matrix A and vector b
A = np.array([[2, 1],
              [1, 3]], dtype=float)

b = np.array([8, 13], dtype=float)

print("Matrix A:")
print(A)

print("\nVector b:")
print(b)
```

```python
# 1. Solve matrix equation Ax = b
solution = np.linalg.solve(A, b)
print("\nSolution of Ax = b:")
print(solution)
```

```python

# 2. Find inverse
inverse = np.linalg.inv(A)
print("\nInverse of A:")
print(inverse)
```

```python
# 3. Find transpose
transpose = A.T
print("\nTranspose of A:")
print(transpose)
```

```python
# 4. Find eigenvalues and eigenvectors
eigenvalues, eigenvectors = np.linalg.eig(A)
print("\nEigenvalues:")
print(eigenvalues)

print("\nEigenvectors:")
print(eigenvectors)
```
