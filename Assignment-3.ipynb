```python
import numpy as np

# Vectorized Operations
a = np.array([1, 2, 3, 4])
b = np.array([10, 20, 30, 40])

print("Addition:", np.add(a, b))
print("Multiplication:", a * b)
```

```python
# Universal Functions
print("Square Root:", np.sqrt(a))
print("Exponential:", np.exp(a))
```

```python
# Broadcasting
c = np.array([1, 2, 3])
d = np.array([[10], [20], [30]])

print("Broadcasting Addition:\n", c + d)
print("Add Scalar:", a + 5)
```
