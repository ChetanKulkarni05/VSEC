```python
import matplotlib.pyplot as plt

x = [1, 2, 3, 4, 5]
y = [10, 20, 15, 25, 30]

plt.figure(figsize=(10, 6))

# Line Plot
plt.subplot(2, 2, 1)
plt.plot(x, y, color='blue', marker='o', label='Line')
plt.title("Line Plot")
plt.legend()
plt.grid(True)
```

```python
# Bar Plot
plt.subplot(2, 2, 2)
plt.bar(x, y, color='green', label='Bar')
plt.title("Bar Plot")
plt.legend()
plt.grid(True)
```

```python
# Scatter Plot
plt.subplot(2, 2, 3)
plt.scatter(x, y, color='red', label='Scatter')
plt.title("Scatter Plot")
plt.legend()
plt.grid(True)
```

```python
# Histogram
plt.subplot(2, 2, 4)
plt.hist(y, bins=5, color='orange', label='Histogram')
plt.title("Histogram")
plt.legend()
plt.grid(True)

plt.tight_layout()
plt.show()
```
