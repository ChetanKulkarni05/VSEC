```python
import pandas as pd
import numpy as np

df = pd.DataFrame({
    "Name": ["Raj", "Amit", "Priya", "Neha"],
    "Marks": [85, np.nan, 90, 70],
    "City": ["Pune", "Mumbai", "Pune", "Mumbai"]
})

print("Data:\n", df)
print("\nMissing Values:\n", df.isnull())
print("\nFill Missing:\n", df.fillna({"Marks": df["Marks"].mean()}))
print("\nDrop Missing:\n", df.dropna())
```

```python
# Replace Data
print("\nReplace:\n", df.replace("Pune", "Delhi"))
```

```python
# Filter Rows and Columns
print("\nFiltered Rows:\n", df[df["Marks"] > 80])
print("\nSelected Columns:\n", df[["Name", "Marks"]])
```

```python
# Sort Values
print("\nSorted Data:\n", df.sort_values("Name"))
```

```python
# Group Values
print("\nGrouped Data:\n", df.groupby("City")["Marks"].mean())
```
