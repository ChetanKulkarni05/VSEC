```python
import numpy as np
```

```python
# 1D Array
a = np.array([10, 20, 30, 40, 50])
print("1D Array:", a)
print("Indexing:", a[2])
print("Slicing:", a[1:4])

# Using arange and linspace
print("Arange:", np.arange(1, 10, 2))
print("Linspace:", np.linspace(0, 10, 5))
```

```python
# 2D Array
b = np.array([[1, 2, 3], [4, 5, 6]])
print("2D Array:\n", b)
print("Indexing:", b[1, 2])
print("Slicing:\n", b[:, 1:3])
```

```python
# Reshaping
c = np.arange(1, 7).reshape(2, 3)
print("Reshaped Array:\n", c)
```

```python
# Mathematical Functions
print("Sum:", np.sum(b))
print("Mean:", np.mean(b))
print("Addition:", b + 2)
print("Multiplication:", b * 2)
```
