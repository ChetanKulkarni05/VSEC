```python

import matplotlib.pyplot as plt

x = [1, 2, 3, 4, 5]
y = [10, 20, 15, 25, 30]

# Line Plot
plt.plot(x, y, marker='o')
plt.xlabel("X Values")
plt.ylabel("Y Values")
plt.title("Line Plot")
plt.show()
```

```python
# Bar Plot
plt.bar(x, y)
plt.xlabel("X Values")
plt.ylabel("Y Values")
plt.title("Bar Plot")
plt.show()
```

```python
# Scatter Plot
plt.scatter(x, y)
plt.xlabel("X Values")
plt.ylabel("Y Values")
plt.title("Scatter Plot")
plt.show()
```

```python
# Histogram
plt.hist(y, bins=5)
plt.xlabel("Values")
plt.ylabel("Frequency")
plt.title("Histogram")
plt.show()
```
